```python
# stack using class 

class stack :
    def __init__(self):
        self.stack=[]

    # def for push

    def push (self,item):
      self.stack.append(item)


    def is_empty (self):
      return len(self.stack)==0

    def pop (self):
     if self.is_empty():
      return"stack is empty"
     return self.stack.pop()

    def peek (self):
      if self.is_empty():
        return"stack is empty"
      return self.stack[-1]

S1 = stack()
S1.push(7)
S1.push(14)
print(S1.pop())
print(S1.pop())
print(S1.pop())
print(S1.peek())

    
```
